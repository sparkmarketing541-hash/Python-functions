## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
# Write your code below
def process_order(price, quantity, discount_rate, tax_rate):
    def calc_subtotal():
        return price * quantity
    def calc_discount(subtotal):
        return subtotal * (discount_rate / 100)
    def calc_tax(amount):
        return amount * (tax_rate / 100)
    def calc_final():
        subtotal = calc_subtotal()
        discount = calc_discount(subtotal)
        discounted_total = subtotal - discount
        tax = calc_tax(discounted_total)
        return discounted_total + tax
    return calc_final()
    
final_amount = process_order(1000, 3, 10, 5)


print("Final Order Amount:", final_amount)
````

## Question 27 — Employee Payroll System

**Problem:**  
Build a small employee payroll system using multiple functions.

**Requirements:**
Your program should contain **separate functions** for:
- Employee information
- Basic salary
- Allowance
- Deduction
- Tax

**Rules:**
- Use arguments and `return` properly.
- Use conditions where required (e.g., tax slabs).
- Do **not** write everything inside one function.
- Display a clean final salary report.

**Focus:** Function design + modular programming

````python
# Write your code below
def get_employee_info(name, emp_id, department):
    return {"name": name, "emp_id": emp_id, "department": department}


def get_basic_salary(grade):
    if grade == "A":
        return 120000
    elif grade == "B":
        return 80000
    else:
        return 50000


def calculate_allowance(basic_salary):
    return basic_salary * 0.20


def calculate_deduction(basic_salary):
    return basic_salary * 0.05


def calculate_tax(gross_salary):
    if gross_salary > 100000:
        return gross_salary * 0.15
    elif gross_salary > 70000:
        return gross_salary * 0.10
    else:
        return gross_salary * 0.05


def calculate_final_salary(basic, allowance, deduction, tax):
    gross = basic + allowance
    net = gross - deduction - tax
    return gross, net


def generate_payroll_report(name, emp_id, department, grade):
    emp = get_employee_info(name, emp_id, department)
    basic = get_basic_salary(grade)
    allowance = calculate_allowance(basic)
    deduction = calculate_deduction(basic)
    gross, temp_net = calculate_final_salary(basic, allowance, deduction, 0)
    tax = calculate_tax(gross)
    _, net = calculate_final_salary(basic, allowance, deduction, tax)

    print("--- Employee Payroll Report ---")
    print(f"Name: {emp['name']}")
    print(f"ID: {emp['emp_id']}")
    print(f"Department: {emp['department']}")
    print(f"Basic Salary: {basic}")
    print(f"Allowance: {allowance}")
    print(f"Deduction: {deduction}")
    print(f"Tax: {tax}")
    print(f"Gross Salary: {gross}")
    print(f"Net Final Salary: {net}")
    print("-------------------------------")


generate_payroll_report("Ali Khan", "EMP101", "IT", "A")
````

## Question 28 — Smart Shopping Billing System

**Problem:**  
Build a shopping billing system that can handle multiple products.

**Requirements:**
- Allow the user (or hard-code) multiple products.
- Calculate:
  - Individual product totals
  - Overall subtotal
  - Discount
  - Tax
  - Final bill
- Use **multiple functions** to separate responsibilities.
- At least one function must use `*args` or a flexible-argument design.
- Display a professional-looking bill.

**Focus:** Real-world application + multiple functions + logic building

````python
def calculate_item_total(price, quantity):
    return price * quantity
def calculate_subtotal(*items):
    return sum(item["total"] for item in items)
def calculate_discount(subtotal):
    return subtotal * 0.10 if subtotal > 5000 else 0
def calculate_tax(amount):
    return amount * 0.05
def generate_bill(*products):
    detailed_items = []
    for name, price, qty in products:
        total = calculate_item_total(price, qty)
        detailed_items.append({"name": name, "price": price, "qty": qty, "total": total})
    subtotal = calculate_subtotal(*detailed_items)
    discount = calculate_discount(subtotal)
    tax = calculate_tax(subtotal - discount)
    final_bill = subtotal - discount + tax
    items_rows = "\n".join(
        f"{item['name']:<15} {item['price']:<8} {item['qty']:<6} {item['total']}"
        for item in detailed_items
    )
    bill_output = (
        f"========================================\n"
        f"          SMART SHOPPING BILL           \n"
        f"========================================\n"
        f"{'Item':<15} {'Price':<8} {'Qty':<6} {'Total'}\n"
        f"----------------------------------------\n"
        f"{items_rows}\n"
        f"----------------------------------------\n"
        f"{'Subtotal:':<28} {subtotal}\n"
        f"{'Discount:':<28} -{discount}\n"
        f"{'Tax (5%):':<28} +{tax}\n"
        f"========================================\n"
        f"{'Final Bill:':<28} {final_bill}\n"
        f"========================================"
    )
    print(bill_output)
generate_bill(
    ("Laptop Bag", 2500, 1),
    ("Wireless Mouse", 1500, 2),
    ("Notebook", 300, 4)
)
````

````python
# Write your code below
def student_info(name, roll_no, student_class):
    return {"name": name, "roll_no": roll_no, "class": student_class}


def calculate_total(*marks):
    return sum(marks)


def calculate_percentage(total, max_per_subject):
    return (total / max_per_subject) * 100


get_grade = lambda pct: (
    "A+" if pct >= 90 else
    "A" if pct >= 80 else
    "B" if pct >= 70 else
    "C" if pct >= 60 else
    "D" if pct >= 50 else "F"
)


def check_status(marks, passing_marks=40):
    for mark in marks:
        if mark < passing_marks:
            return "Fail"
    return "Pass"


def generate_result_card(name, roll_no, student_class, subjects_scores, max_marks_per_subject=100):
    info = student_info(name, roll_no, student_class)
    marks_list = [score for _, score in subjects_scores]
    total = calculate_total(*marks_list)
    total_possible = len(marks_list) * max_marks_per_subject
    percentage = calculate_percentage(total, total_possible)
    grade = get_grade(percentage)
    status = check_status(marks_list)

    subject_lines = "\n".join(
        f"{subj:<18} {score} / {max_marks_per_subject}"
        for subj, score in subjects_scores
    )

    card = (
        f"========================================\n"
        f"          STUDENT RESULT CARD           \n"
        f"========================================\n"
        f"Name:       {info['name']}\n"
        f"Roll No:    {info['roll_no']}\n"
        f"Class:      {info['class']}\n"
        f"----------------------------------------\n"
        f"{subject_lines}\n"
        f"----------------------------------------\n"
        f"Total:      {total} / {total_possible}\n"
        f"Percentage: {percentage:.2f}%\n"
        f"Grade:      {grade}\n"
        f"Status:     {status}\n"
        f"========================================"
    )

    print(card)


generate_result_card(
    "Ayesha Tariq",
    "104",
    "10th",
    [("Mathematics", 88), ("Physics", 79), ("Chemistry", 84), ("English", 91), ("Computer", 95)]
)
````

````python
# Write your code below
def get_customer_info(name, email, city="Karachi", **extra_details):
    customer = {"name": name, "email": email, "city": city}
    customer.update(extra_details)
    return customer


def build_product_catalog():
    return {
        "P01": {"name": "Mechanical Keyboard", "price": 8500},
        "P02": {"name": "Gaming Mouse", "price": 4200},
        "P03": {"name": "USB-C Hub", "price": 3100},
        "P04": {"name": "Desk Mat", "price": 1500},
    }


def select_products(catalog, *order_items):
    selected = []
    for product_id, qty in order_items:
        if product_id in catalog and qty > 0:
            item = catalog[product_id]
            selected.append({
                "id": product_id,
                "name": item["name"],
                "price": item["price"],
                "qty": qty,
            })
    return selected


def calculate_item_totals(items):
    calc_line = lambda price, qty: price * qty
    totals = []
    for item in items:
        line_total = calc_line(item["price"], item["qty"])
        totals.append({**item, "total": line_total})
    return totals


def calculate_subtotal(priced_items):
    return sum(item["total"] for item in priced_items)


def calculate_discount(subtotal, promo_code=None):
    if promo_code == "SAVE10" or subtotal >= 15000:
        return subtotal * 0.10
    elif subtotal >= 8000:
        return subtotal * 0.05
    return 0.0


def calculate_tax(amount, tax_rate=0.08):
    return amount * tax_rate


def calculate_delivery(city, discounted_amount):
    def base_city_fee(target_city):
        if target_city.lower() in ("karachi", "lahore", "islamabad"):
            return 250.0
        return 450.0

    if discounted_amount >= 20000:
        return 0.0
    return base_city_fee(city)


def calculate_final_bill(subtotal, discount, tax, delivery):
    return (subtotal - discount) + tax + delivery


def display_order_summary(customer, priced_items, subtotal, discount, tax, delivery, final_bill):
    item_lines = "\n".join(
        f"{item['name']:<22} {item['price']:<8} {item['qty']:<5} {item['total']}"
        for item in priced_items
    )

    summary = (
        f"======================================================\n"
        f"               E-COMMERCE ORDER SUMMARY               \n"
        f"======================================================\n"
        f"Customer: {customer['name']} ({customer['email']})\n"
        f"City:     {customer['city']}\n"
        f"------------------------------------------------------\n"
        f"{'Product':<22} {'Price':<8} {'Qty':<5} {'Total'}\n"
        f"------------------------------------------------------\n"
        f"{item_lines}\n"
        f"------------------------------------------------------\n"
        f"{'Subtotal:':<42} {subtotal:.2f}\n"
        f"{'Discount:':<42} -{discount:.2f}\n"
        f"{'Tax:':<42} +{tax:.2f}\n"
        f"{'Delivery:':<42} +{delivery:.2f}\n"
        f"======================================================\n"
        f"{'Final Bill:':<42} {final_bill:.2f}\n"
        f"======================================================"
    )

    print(summary)


def run_order_system():
    customer = get_customer_info("Bilal Ahmed", "bilal@example.com", "Lahore", phone="0300-1234567")
    catalog = build_product_catalog()
    selected = select_products(catalog, ("P01", 1), ("P02", 2), ("P04", 1))
    priced_items = calculate_item_totals(selected)

    subtotal = calculate_subtotal(priced_items)
    discount = calculate_discount(subtotal, promo_code="SAVE10")
    taxable_amount = subtotal - discount
    tax = calculate_tax(taxable_amount)
    delivery = calculate_delivery(customer["city"], taxable_amount)
    final_bill = calculate_final_bill(subtotal, discount, tax, delivery)

    display_order_summary(customer, priced_items, subtotal, discount, tax, delivery, final_bill)


run_order_system()
````
